```python
import pandas as pd
from sqlalchemy import create_engine

# Define your database connection string
db_connection_str = 'mysql+pymysql://root:root@localhost:3306/SQL_Employee_Project'

# Create a SQLAlchemy engine
engine = create_engine(db_connection_str)

# Load the Excel file
xls = pd.ExcelFile('C:\\Users\\Lenovo\\OneDrive\\Desktop\\Data Analytics Kapil\\SQL\\Project\\SQL_Project\\SQL_Project_Data.xlsx')

# Loop through all sheet names in the Excel file
for sheet_name in xls.sheet_names:
    # Read each sheet into a pandas DataFrame
    df = pd.read_excel('C:\\Users\\Lenovo\\OneDrive\\Desktop\\Data Analytics Kapil\\SQL\\Project\\SQL_Project\\SQL_Project_Data.xlsx', sheet_name=sheet_name)

    # Ensure column names are strings and replace spaces with underscores
    df.columns = df.columns.astype(str).str.replace(' ', '_')

    # Print statement for debugging
    print(f"Processing sheet: {sheet_name}")
    print(df.head())  # Print first few rows to confirm data is read correctly

    # Convert the DataFrame to a SQL table
    try:
        df.to_sql(sheet_name, con=engine, if_exists='replace', index=False)
    except Exception as e:
        print(f"Error processing sheet '{sheet_name}': {e}")

# Close the SQLAlchemy engine
engine.dispose()
```
